```python
import os

path = "data/dataset1"

classes = ["cardboard", "glass", "metal", "paper", "plastic", "trash"]

print("📊 DATASET1 STATS\n")

total = 0

for c in classes:
    folder = os.path.join(path, c)

    if os.path.exists(folder):
        count = len(os.listdir(folder))
        total += count

        print(f"{c}: {count} images")

print("\nTOTAL:", total, "images")
```
